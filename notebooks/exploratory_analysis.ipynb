# Customer Churn & Revenue Intelligence Platform
## Exploratory Data Analysis & Predictive Modeling

**Objective:** Analyze customer churn dynamics, identify high-risk cohorts, quantify revenue exposure, and evaluate machine learning models (Logistic Regression, Decision Trees, Random Forests).

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_data
from src.preprocessing import preprocess_data
from src.feature_engineering import engineer_features
from src.churn_model import train_models

sns.set_theme(style="darkgrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
```

### 1. Data Ingestion & Preprocessing Pipeline

```python
df_raw = load_data()
print(f"Raw dataset shape: {df_raw.shape}")
df_clean = preprocess_data(df_raw)
df_eng = engineer_features(df_clean)
print(f"Engineered dataset shape: {df_eng.shape}")
df_eng.head()
```

### 2. Churn Distribution & Core Baseline Metrics

```python
churn_counts = df_eng['churn'].value_counts(normalize=True) * 100
print(f"Overall Churn Rate: {churn_counts['Yes']:.2f}%")

plt.figure(figsize=(6, 4))
sns.countplot(data=df_eng, x='churn', palette=['#1BC5BD', '#F64E60'])
plt.title('Overall Customer Churn Distribution')
plt.show()
```

### 3. Contract Type vs. Churn Rate

```python
contract_churn = df_eng.groupby('contract')['churn_flag'].mean() * 100
print(contract_churn)

plt.figure(figsize=(8, 4))
contract_churn.plot(kind='bar', color='#6C63FF')
plt.title('Churn Rate by Contract Type (%)')
plt.ylabel('Churn Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

### 4. Monthly Charges and Tenure Dynamics

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_eng, x='contract', y='monthly_charges', hue='churn', palette=['#1BC5BD', '#F64E60'])
plt.title('Monthly Charges by Contract Type and Churn Status')
plt.show()
```

### 5. Machine Learning Model Training & Comparative Evaluation

```python
all_metrics, best_pipe, metadata = train_models(df_eng)
metrics_df = pd.DataFrame(all_metrics)[['model_name', 'accuracy', 'precision', 'recall', 'f1', 'roc_auc']]
print("Model Evaluation Benchmark:")
print(metrics_df.to_string(index=False))
```
